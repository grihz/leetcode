Implement **pow(x, n)**, which calculates x raised to the power n (i.e., xn).

**Example 1:**
> Input: x = 2.00000, n = 10
> Output: 1024.00000

**Example 2:**
> Input: x = 2.10000, n = 3
> Output: 9.26100

**Example 3:**
> Input: x = 2.00000, n = -2
> Output: 0.25000
> Explanation: 2-2 = 1/22 = 1/4 = 0.25

**Constraints:**
+ -100.0 < x < 100.0
+ -231 <= n <= 231-1
+ n is an integer.
+ Either x is not zero or n > 0.
+ -104 <= xn <= 104


```python
class Solution:
    def myPow(self, x: float, n: int) -> float:
        result = x
        if n == 0:
            return float(1)
        for i in range (1, abs(n)):
            result *= x
        return result if n > 0 else 1/result
```

```python
class Solution:
    def myPow(self, x: float, n: int) -> float:
        if n == 0:
            return float(1)
        m = abs(n)
        result = x
        i = 1
        while 2 * i <= m:
            i *= 2
            result *= result
        result = result * self.myPow(x, m - i)
        return result if n > 0 else 1/result
```
