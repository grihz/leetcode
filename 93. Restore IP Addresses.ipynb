A **valid IP address** consists of exactly four integers separated by single dots. Each integer is between 0 and 255 (inclusive) and cannot have leading zeros.

+ For example, `"0.1.2.201"` and `"192.168.1.1"` are valid IP addresses, but `"0.011.255.245"`, `"192.168.1.312"` and `"192.168@1.1"` are invalid IP addresses.

Given a string s containing only digits, return all possible valid IP addresses that can be formed by inserting dots into s. You are not allowed to reorder or remove any digits in s. You may return the valid IP addresses in any order.


**Example 1:**
> Input: s = "25525511135"
> Output: ["255.255.11.135","255.255.111.35"]

**Example 2:**
> Input: s = "0000"
> Output: ["0.0.0.0"]

**Example 3:**
> Input: s = "101023"
> Output: ["1.0.10.23","1.0.102.3","10.1.0.23","10.10.2.3","101.0.2.3"]



**Constraints:**
+ 1 <= s.length <= 20
+ s consists of digits only.

```python
from collections import deque

class Solution_1:
    def restoreIpAddresses(self, s: str) -> list[str]:
        result = []
        dq = deque()
        dq.append(self.buildOctet(s, 0))
        length = len(s)
        for i in range(length - 3):
            octet1 = dq.popleft()
            for k in range(3):
                o1 = octet1[k]
                if o1 is None:
                    break
                if len(dq) < (k + 1):
                    dq.append(self.buildOctet(s, i + k + 1))
                octet2 = dq[k]
                for l in range(3):
                    o2 = octet2[l]
                    if o2 is None:
                        break
                    if len(dq) < (k + l + 2):
                        dq.append(self.buildOctet(s, i + k + l + 2))
                    octet3 = dq[k + l +1]
                    for m in range(3):
                        o3 = octet3[m]
                        if o3 is None:
                            break
                        if len(dq) < (k + m + l + 3):
                            dq.append(self.buildOctet(s, i + k + m + l + 3))
                        octet4 = dq[k + m + l + 2]
                        for n in range(3):
                            o4 = octet4[n]
                            if o4 is None:
                                break
                            result.append(f'{o1}.{o2}.{o3}.{o4}')
        return result


    def buildOctet(self, s: str, n: int) -> tuple[str | None, str | None, str | None]:
        length = len(s)
        if n > (length - 1):
            return None, None, None
        if n == (length - 1):
            return s[n], None, None
        if n == (length - 2):
            return s[n], s[n:n+2], None
        return s[n], s[n:n + 2], s[n:n + 3]
```

```python

class Solution:
    def restoreIpAddresses(self, s: str) -> list[str]:
        result = []
        length = len(s)
        if length < 3 or length > 12:
            return []
        if length == 12:
            o1 = s[0:3]
            o2 = s[3:6]
            o3 = s[6:9]
            o4 = s[9:12]
            if self.isCorrectOctet(o1) and self.isCorrectOctet(o2) and self.isCorrectOctet(o3) and self.isCorrectOctet(o4):
                return [f'{o1}.{o2}.{o3}.{o4}']
            else:
                return []
        if length == 4:
            o1 = s[0]
            o2 = s[1]
            o3 = s[2]
            o4 = s[3]
            if self.isCorrectOctet(o1) and self.isCorrectOctet(o2) and self.isCorrectOctet(o3) and self.isCorrectOctet(o4):
                return [f'{o1}.{o2}.{o3}.{o4}']
            else:
                return []
        for i in range(3):
            o1 = s[0:i+1]
            if not self.isCorrectOctet(o1):
                break
            sub = s[i+1:]
            if len(sub) < 3:
                break
            for j in range(3):
                o2 = s[i+1:i+j+2]
                if not self.isCorrectOctet(o2):
                    break
                sub = s[i+j+2:]
                if len(sub) < 2:
                    break
                for k in range(3):
                    o3 = s[i+j+2:i+j+k+3]
                    if not self.isCorrectOctet(o3):
                        break
                    sub = s[i+j+k+3:]
                    if len(sub) == 0:
                        break
                    if len(sub) >= 4:
                        continue
                    if not self.isCorrectOctet(sub):
                        continue
                    result.append(f'{o1}.{o2}.{o3}.{sub}')
        return result


    def isCorrectOctet(self, s: str) -> bool:
        length = len(s)
        if length > 3 or length == 0:
            return False
        if length == 1:
            return True
        if length == 2 and s[0] != '0' :
            return True
        if length == 3 and (s[0] == '1' or (s[0] == '2' and ( s[1] in ('0', '1', '2', '3', '4') or (s[1] == '5' and s[2] in ('0', '1', '2', '3', '4', '5'))))):
            return True
        return False
```

```python
import unittest
from parameterized import parameterized

class RestoreIpTestNotebook(unittest.TestCase):

    def setUp(self):
        self.s = Solution()


    def assertIpAddressesAreEqual(self, expected_ip_addresses: list[str], actual_ip_addresses: list[str]):
        if len(expected_ip_addresses) != len(actual_ip_addresses):
            self.fail('Ip addresses did not match. \n Expected ip addresses: {!r} \n Actual ip addresses: {!r}'.format(expected_ip_addresses,actual_ip_addresses))
        for n1 in expected_ip_addresses:
            found: str | None = None
            for n2 in actual_ip_addresses:
                if n1 == n2:
                    found = n2
                    break
            if found is None:
                self.fail('Ip addresses list did not match.\n Expected ip addresses list: {!r} \n Actual ip addresses list: {!r}'.format(expected_ip_addresses,actual_ip_addresses))

    @parameterized.expand([
        ("101023", ['1.0.10.23', '1.0.102.3', '10.1.0.23', '10.10.2.3', '101.0.2.3']),
        ("25525511135", ['255.255.11.135', '255.255.111.35']),
        ("0000", ['0.0.0.0']),
        ("00000", [])
    ])
    def test_restored_ip_addresses(self, source, expected):
        self.assertIpAddressesAreEqual(self.s.restoreIpAddresses(source), expected)


unittest.main(argv=[''], verbosity=2, exit=False)
```
