Given a **valid** parentheses string s, return the nesting depth of s. The nesting depth is the maximum number of nested parentheses.

**Example 1:**
> Input: s = "(1+(2*3)+((8)/4))+1"
> Output: 3
> Explanation:
> Digit 8 is inside of 3 nested parentheses in the string.

**Example 2:**
> Input: s = "(1)+((2))+(((3)))"
> Output: 3
> Explanation:
> Digit 3 is inside of 3 nested parentheses in the string.


**Example 3:**
> Input: s = "()(())((()()))"
> Output: 3

**Constraints:**
+ 1 <= s.length <= 100
+ consists of digits 0-9 and characters '+', '-', '*', '/', '(', and ')'.
+ is guaranteed that parentheses expression s is a VPS.


```python
from collections import deque


class Solution:
    def maxDepth(self, s: str) -> int:
        max_depth = 0
        brackets = deque()
        length = len(s)
        i = 0
        while (i + len(brackets)) < length:
            if s[i] == '(':
                brackets.append(i)
            if s[i] == ')':
                depth = len(brackets)
                if depth > max_depth:
                    max_depth = depth
                brackets.pop()
            i += 1
        depth = len(brackets)
        if depth > max_depth:
            max_depth = depth
        return max_depth
```

```python
print(Solution().maxDepth("()(())((()()))"))
```
