Given a string containing just the characters *'('* and *')'*, return the length of the longest valid (well-formed) parentheses substring.

**Example 1:**
> Input: s = "(()"
> Output: 2 Explanation:
> The longest valid parentheses substring is "()".

**Example 2:**
 > Input: s = ")()())"
 > Output: 4
 > Explanation: The longest valid parentheses substring is "()()".

**Example 3:**
> Input: s = ""
> Output: 0

**Constraints:**
  + 0 <= s.length <= 3 * 104
  + s[i] is '(', or ')'.

```python
from collections import deque

class Solution:

    def longestValidParentheses(self, s: str) -> int:
        j = 0
        length = len(s)
        bracket_number = deque()
        max_valid_parentheses = 0
        current_valid_parentheses = 0
        while j < length:
            if len(bracket_number) == 0 and s[j] == ')':
                current_valid_parentheses = 0
                j += 1
                continue
            if s[j] == '(':
                bracket_number.append(current_valid_parentheses + 1)
                current_valid_parentheses = 0
                j += 1
                continue
            current_valid_parentheses += bracket_number.pop() + 1
            max_valid_parentheses = max(max_valid_parentheses, current_valid_parentheses)
            j += 1
        return max_valid_parentheses

```

```python

class Solution:

    def longestValidParentheses(self, s: str) -> int:
        j = 0
        length = len(s)
        s += "("
        bracket_number = deque()
        max_valid_parentheses = 0
        current_valid_parentheses = 0
        while j < length:
            if len(bracket_number) == 0 and s[j] == ')':
                max_valid_parentheses = max(max_valid_parentheses, current_valid_parentheses)
                current_valid_parentheses = 0
                while s[j] != '(':
                    j += 1
                else:
                    continue
            if s[j] == '(':
                max_valid_parentheses = max(max_valid_parentheses, current_valid_parentheses)
                while s[j] == '(' and j < length :
                    bracket_number.append(current_valid_parentheses + 1)
                    current_valid_parentheses = 0
                    j += 1
                else:
                    continue
            current_valid_parentheses += bracket_number.pop() + 1
            j += 1
        max_valid_parentheses = max(max_valid_parentheses, current_valid_parentheses)
        return max_valid_parentheses
```
